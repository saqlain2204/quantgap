# Make a quantgap pack on a free Kaggle GPU

A **pack** holds the original model's answers on a fixed set of texts. You make it once, then everyone can score quants of that model without running the big model.

Before you run:
1. Notebook settings: **Accelerator = GPU T4 x2**, **Internet = on**.
2. Add a Kaggle secret named `HF_TOKEN` (a Hugging Face token with write access) if you want to upload.
3. Check the model's license allows sharing its outputs. Do not publish a pack for a model whose license forbids it.

Packs for models above about 32B parameters do not fit on free hardware.

````python
MODEL = "Qwen/Qwen2.5-0.5B-Instruct"   # change me
REVISION = ""                           # full commit sha from the model page (recommended)
LICENSE = "apache-2.0"                  # the model's license
PUSH_TO = ""                            # e.g. "your-org/Qwen2.5-0.5B-Instruct-pack", empty = do not upload
````

````python
!pip install -q "quantgap[pack]"
````

````python
import os, subprocess
try:
    from kaggle_secrets import UserSecretsClient
    os.environ["HF_TOKEN"] = UserSecretsClient().get_secret("HF_TOKEN")
except Exception:
    print("No HF_TOKEN secret found: upload will not work.")

cmd = ["quantgap", "pack", MODEL, "--license", LICENSE, "--out", "/kaggle/working/pack"]
if REVISION:
    cmd += ["--revision", REVISION]
if PUSH_TO:
    cmd += ["--push", PUSH_TO]
print(" ".join(cmd))
subprocess.run(cmd, check=True)
````

````python
!quantgap packs show /kaggle/working/pack
!du -sh /kaggle/working/pack
````

Download `/kaggle/working/pack` from the Output tab, or use the uploaded dataset with:

```
quantgap score <your quant> --pack your-org/<dataset>
```
